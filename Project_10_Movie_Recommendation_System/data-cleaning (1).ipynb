```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from ydata_profiling import ProfileReport
import os
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
DATA_PATH = "/kaggle/input/datasets/mhuzaifayounas/dataset/data/ml-latest-small"
print(f"Data path : {DATA_PATH}")
```

```python
ratings_df = pd.read_csv(f'{DATA_PATH}/ratings.csv')
print(f"\nratings.csv loaded: {len(ratings_df):,} rows, {len(ratings_df.columns)} columns")

movies_df = pd.read_csv(f'{DATA_PATH}/movies.csv')
print(f"\nmovies.csv loaded : {len(movies_df):,} rows, {len(movies_df.columns)} columns.")

tags_df = pd.read_csv(f'{DATA_PATH}/tags.csv')
print(f"\nmovies.csv loaded : {len(movies_df):,} rows, {len(movies_df.columns)} columns.")

links_df = pd.read_csv(f'{DATA_PATH}/links.csv')
print(f"\nmovies.csv loaded : {len(movies_df):,} rows, {len(movies_df.columns)} columns.")

```
